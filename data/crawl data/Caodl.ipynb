```python
import requests
from bs4 import BeautifulSoup
import re
import json
import time

# Danh sách link cần cào
list_urls = [
    "https://vneconomy.vn/cpi-thang-1-2021-tang-006.htm",
    "https://vneconomy.vn/cpi-thang-2-2021-tang-152-muc-tang-cao-nhat-cua-thang-2-trong-8-nam.htm",
    "https://vneconomy.vn/nguyen-nhan-nao-khien-cpi-thang-3-giam-027.htm",
    "https://vneconomy.vn/cpi-thang-4-giam-co-phai-la-nghich-ly.htm",
    "https://vneconomy.vn/giua-bao-gia-nguyen-lieu-vi-sao-cpi-5-thang-van-thap-nhat-5-nam.htm",
    "https://vneconomy.vn/cpi-6-thang-thap-nhat-trong-5-nam-khong-the-chu-quan.htm",
    "https://vneconomy.vn/gia-luong-thuc-hang-hoa-thiet-yeu-tang-day-cpi-thang-7-tang-0-62.htm",
    "https://vneconomy.vn/gia-luong-thuc-thuc-pham-tang-day-cpi-thang-8-tang-0-25.htm",
    "https://vneconomy.vn/mien-giam-hoc-phi-giam-gia-dien-khien-cpi-thang-9-giam-0-62.htm",
    "https://vneconomy.vn/cpi-thang-10-2021-giam-0-2-so-voi-thang-9-tang-1-81-cung-ky.htm",
    "https://vneconomy.vn/gia-xang-dau-leo-thang-keo-cpi-thang-11-tang.htm",
    "https://vneconomy.vn/cpi-binh-quan-nam-2021-tang-1-84-muc-thap-nhat-ke-tu-2016.htm",

    "https://vneconomy.vn/nhu-cau-sam-tet-cao-day-cpi-thang-dau-nam-tang.htm",
    "https://vneconomy.vn/cpi-thang-02-2022-tang-do-tac-dong-tu-gia-xang-dau-va-luong-thuc.htm",
    "https://vneconomy.vn/chi-so-gia-tieu-dung-cpi-thang-3-tang-0-7.htm",
    "https://vneconomy.vn/cpi-thang-4-tang-nhe-o-muc-0-18.htm",
    "https://vneconomy.vn/luong-huu-duoc-dieu-chinh-theo-cpi-va-toc-do-tang-truong-kinh-te.htm",
    "https://vneconomy.vn/cpi-thang-6-chiu-tac-dong-manh-tu-gia-xang-dau.htm",
    "https://vneconomy.vn/luong-thuc-xang-dau-day-cpi-thang-7-tang-nhe.htm",
    "https://vneconomy.vn/gia-xang-dau-lien-tuc-giam-cpi-thang-8-chi-tang-0-005.htm",
    "https://vneconomy.vn/gia-nguyen-vat-lieu-dau-vao-cao-day-cpi-thang-9-tang-0-4.htm",
    "https://vneconomy.vn/cpi-thang-10-tang-0-15-lam-phat-van-trong-tam-kiem-soat.htm",
    "https://vneconomy.vn/cpi-thang-11-2022-tang-0-39.htm",
    "https://vneconomy.vn/xang-dau-giam-gia-keo-cpi-thang-12-giam-nhe.htm",

    "https://vneconomy.vn/nhu-cau-mua-sam-tet-nguyen-dan-cao-da-day-cpi-thang-1-tang-0-52.htm",
    "https://vneconomy.vn/gia-xang-dau-day-cpi-thang-2-tang.htm",
    "https://vneconomy.vn/xang-dau-luong-thucha-nhiet-keo-cpi-thang-3-giam-0-23.htm",
    "https://vneconomy.vn/nhieu-nhom-hang-ha-nhiet-cpi-thang-4-giam-0-34.htm",
    "https://vneconomy.vn/luong-thuc-thuc-pham-gia-dien-day-cpi-thang-5-tang-0-01.htm",
    "https://vneconomy.vn/cpi-thang-6-tang-cao-do-tac-dong-cua-gia-dien-va-gia-thuc-pham.htm",
    "https://vneconomy.vn/nhu-cau-tieu-dung-tang-cao-vao-mua-du-lich-day-cpi-thang-7-tang-0-45.htm",
    "http://vneconomy.vn/cpi-thang-8-tang-theo-gia-xang-dau-gao.htm",
    "https://vneconomy.vn/cpi-thang-9-tiep-tuc-chiu-ap-luc-do-tac-dong-tu-nhieu-mat-hang-tang-gia.htm",
    "https://vneconomy.vn/cpi-thang-10-tang-nhe-lam-phat-van-trong-tam-kiem-soat.htm",
    "https://vneconomy.vn/gia-vang-the-gioi-tang-manh-sau-tin-cpi-my.htm",
    "https://vneconomy.vn/cpi-nam-2023-tang-3-25-dat-muc-tieu-quoc-hoi-de-ra.htm",

    "https://vneconomy.vn/gia-thuoc-va-dich-vu-y-te-tang-cao-day-cpi-thang-1-2024-tang-0-31.htm",
    "https://vneconomy.vn/nhu-cau-mua-sam-tet-gia-gao-xang-dau-day-cpi-thang-2-tang-1-04.htm",
    "https://vneconomy.vn/nhu-cau-tieu-dung-sau-tet-giam-keo-cpi-thang-3-giam-0-23.htm",
    "https://vneconomy.vn/cpi-thang-4-tang-do-tac-dong-cua-gia-xang-dau.htm",
    "https://vneconomy.vn/cpi-thang-5-2024-tang-do-tac-dong-tu-gia-thit-lon-gia-dien.htm",
    "https://vneconomy.vn/cpi-my-yeu-hon-du-bao-kha-nang-fed-ha-lai-suat-tang-len.htm",
    "https://vneconomy.vn/cpi-thang-7-tang-0-48-lam-phat-co-ban-tang-0-36-so-voi-thang-truoc.htm",
    "https://vneconomy.vn/cpi-thang-8-2024-co-ban-on-dinh.htm",
    "https://vneconomy.vn/cpi-thang-9-2024-tang-0-29-lam-phat-co-ban-tang-0-27-so-voi-thang-truoc.htm",
    "https://vneconomy.vn/gia-luong-thuc-thuc-pham-xang-dau-tang-day-cpi-thang-10-2024-tang-0-33.htm",
    "http://vneconomy.vn/cpi-thang-11-tang-0-13-do-tac-dong-tu-gia-dien-gia-nha-o-thue-tang.htm",
    "https://vneconomy.vn/cpi-nam-2024-tang-3-63-lam-phat-duoc-kiem-soat-hieu-qua.htm",
    
    "https://vneconomy.vn/nhu-cau-tieu-dung-dip-tet-tang-cao-day-cpi-thang-1-2025-tang-gan-1.htm",
    "https://vneconomy.vn/gia-thit-lon-dich-vu-giao-thong-keo-cpi-thang-2-tang-0-34-so-voi-thang-truoc.htm",
    "https://vneconomy.vn/gia-xang-dau-gao-giam-theo-gia-the-gioi-keo-cpi-thang-3-giam-0-03.htm",
    "https://vneconomy.vn/gia-nha-o-thue-gia-thuc-pham-day-cpi-thang-4-tang-0-07.htm",
    "https://vneconomy.vn/cpi-thang-5-tang-0-16.htm",
    "https://vneconomy.vn/cpi-binh-quan-6-thang-tang-3-27-lam-phat-van-trong-tam-kiem-soat.htm",
    "https://vneconomy.vn/9-nhom-hang-hoa-dich-vu-tang-gia-keo-cpi-thang-7-tang-0-11.htm",
    "https://vneconomy.vn/cpi-thang-8-tang-nhe.htm",
    "https://vneconomy.vn/cpi-thang-9-tang-042-lam-phat-van-trong-tam-kiem-soat.htm",
    "https://vneconomy.vn/mua-lu-keo-dai-gia-thuc-pham-day-cpi-thang-10-tang-nhe.htm",
    "https://vneconomy.vn/gia-thuc-pham-tang-sau-bao-day-cpi-thang-112025-tang-045.htm",
    "https://vneconomy.vn/cpi-nam-2025-tang-331-dat-muc-tieu-quoc-hoi-de-ra.htm"
    "https://vneconomy.vn/nhu-cau-tieu-dung-dip-tet-binh-ngo-day-cpi-thang-2-tang-114.htm",

    
]

def get_safe_text(soup, tag_name, class_name=None):
    """Hàm lấy text an toàn, tránh lỗi khi không tìm thấy thẻ"""
    if class_name:
        tag = soup.find(tag_name, class_=class_name)
    else:
        tag = soup.find(tag_name)
    return tag.get_text(strip=True) if tag else ""

def crawl_cpi_full_info(url):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }
    
    try:
        response = requests.get(url, headers=headers, timeout=10)
        if response.status_code != 200: return None
        
        soup = BeautifulSoup(response.content, 'html.parser')

        # --- 1. LẤY TIÊU ĐỀ (Yêu cầu của bạn) ---
        title = get_safe_text(soup, 'h1', 'detail-title') or get_safe_text(soup, 'h1')

        # --- 2. LẤY NGÀY THÁNG ---
        date_str = "N/A"
        # Ưu tiên lấy từ Meta data (chuẩn xác nhất)
        meta_date = soup.find('meta', property='article:published_time') or soup.find('meta', itemprop='datePublished')
        if meta_date:
            raw_date = meta_date.get('content', '')
            match_d = re.search(r'(\d{4})-(\d{2})-(\d{2})', raw_date)
            if match_d:
                date_str = f"{match_d.group(3)}/{match_d.group(2)}/{match_d.group(1)}"
        else:
            # Nếu không có meta, tìm trong giao diện
            date_display = get_safe_text(soup, 'div', 'detail-time')
            match_d2 = re.search(r'\d{1,2}/\d{1,2}/\d{4}', date_display)
            if match_d2: date_str = match_d2.group(0)

        # --- 3. LẤY SỐ LIỆU & XU HƯỚNG ---
        # Lấy nội dung để quét
        sapo_text = get_safe_text(soup, 'h2', 'detail-sapo')
        content_text = get_safe_text(soup, 'div', 'detail-content') or get_safe_text(soup, 'div', 'content')
        full_text = f"{title}. {sapo_text} {content_text}"

        value_change = None
        trend = "Không rõ"
        
        # Regex tìm: Từ khóa ... Tăng/Giảm ... Con số %
        pattern = r'(?i)(CPI|lạm phát|chỉ số giá tiêu dùng).{0,100}?(tăng|giảm|ổn định).{0,50}?(\d+[.,]\d+)\s*%'
        match = re.search(pattern, full_text)
        
        if match:
            # Lấy số %
            if match.group(3):
                value_change = match.group(3).replace(',', '.')
            
            # Lấy xu hướng
            trend_keyword = match.group(2).lower()
            if "tăng" in trend_keyword: trend = "Tăng"
            elif "giảm" in trend_keyword: trend = "Giảm"
            elif "ổn định" in trend_keyword: trend = "Đi ngang"
        else:
            # Fallback: Tìm trong tiêu đề nếu nội dung không thấy
            match_title = re.search(r'(\d+[.,]\d+)\s*%', title)
            if match_title:
                value_change = match_title.group(1).replace(',', '.')
                if "tăng" in title.lower(): trend = "Tăng"
                elif "giảm" in title.lower(): trend = "Giảm"

        # --- 4. TẠO DICTIONARY JSON (Đúng yêu cầu) ---
        return {
            "tieu_de": title,                  # Tiêu đề bài báo
            "ngay_thang": date_str,            # Ngày đăng
            "xu_huong_bien_dong": trend,       # Tăng hay Giảm
            "muc_tang_giam": value_change,     # Con số cụ thể (VD: 0.34)
            "don_vi": "%",
            "url": url
        }

    except Exception as e:
        print(f"Lỗi link {url}: {e}")
        return None

# --- CHẠY VÀ LƯU FILE ---
results = []
print("Đang xử lý dữ liệu...\n")

for link in list_urls:
    data = crawl_cpi_full_info(link)
    if data:
        results.append(data)
        print(f"✅ Đã lấy: {data['tieu_de'][:50]}... | {data['xu_huong_bien_dong']} {data['muc_tang_giam']}%")
    time.sleep(1)

# Lưu file JSON
filename = 'du_lieu_cpi_hoan_chinh.json'
with open(filename, 'w', encoding='utf-8') as f:
    json.dump(results, f, ensure_ascii=False, indent=4)

print(f"\n🎉 Đã lưu thành công vào file '{filename}'")
```
